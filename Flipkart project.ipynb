### Key Insights Summary

1. Catalog Composition
Clothing and Jewellery dominate the catalog, together accounting for nearly half of all ~19,920 products. The remaining 30 categories form a long tail, with many having fewer than 100 listings.

2. Pricing Patterns
Prices are heavily right-skewed (median ₹1,039 vs. mean ₹2,979), with a secondary cluster of premium items (₹30k-50k) likely driven by jewelry/watches. Furniture, Automotive, and Watches command the highest typical prices; Home Improvement, Baby Care, and Pet Supplies are consistently the cheapest.

3. Discounting Strategy
Discounts show two distinct spikes — ~12% of products have no discount at all, and another ~11% are discounted at almost exactly 50%, suggesting deliberate "round number" pricing rather than organic demand-based discounting. Discount intensity also varies sharply by category: Beauty and Personal Care is discounted far less (median ~15%) than Automotive or Mobiles & Accessories (median ~55-60%).

4. Ratings Are Sparse and Polarized
Only ~9.2% of products (1,839 of 19,920) have any customer rating at all. Among rated products, ratings follow a J-shaped distribution — heavily clustered at 5.0 and, to a lesser extent, 1.0 — reflecting well-documented reviewer bias where moderate experiences go unreported.

5. Rating Is Independent of Price and Discount
Neither retail price nor discount percentage shows a meaningful correlation with product rating (correlation coefficients of 0.05 and -0.002 respectively). Expensive products aren't rated better; heavily discounted products aren't rated worse. Rating appears to be driven by product/brand-specific factors not captured in these fields.

6. FK Advantage Products Behave Differently
Only ~3.9% of products carry the "FK Advantage" label. These products are priced similarly to regular listings but are discounted noticeably less (avg ~28% vs ~41%), and lack the extreme high-price outliers seen elsewhere — consistent with the label functioning as a trust/fulfillment signal rather than a pricing tier.

7. Brand Matters — Somewhat
Among brands with sufficient rating volume, quality varies widely, from JDX/Bosch (~4.95 avg) down to Clovia (3.25) — which is actually below the 3.77 average for unbranded ("Unknown") products. Brand recognition alone doesn't guarantee better ratings.

8. Data Quality Note
~29% of products had no brand listed (filled as "Unknown"), and some brand entries were found to be non-brand values (sizes/colors like "Regular," "Slim," "Black") — a data entry inconsistency in the original source worth flagging as a limitation.
